```python
cus_df=spark.sql("select * from samples.bakehouse.sales_customers where continent='Asia'").display()
```

```python
cus_df=cus_df.drop("continent","phone_number","email_address")
cus_df.display()
```

```python
result_df= (
    cus_df
    .withColumnRenamed("customerID","cus_ID")
    .withColumnRenamed("first_name","name")
    .withColumnRenamed("address","info")
    
)

result_df.display()
```
