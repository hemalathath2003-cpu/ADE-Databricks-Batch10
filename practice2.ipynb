```python
from pyspark.sql import functions as F
```

```python
ego_df=spark.read.option("multiline","true").json("/Volumes/qc_databricks_batch_10/default/input_files/cars.json")
ego_df.display()

```

```python
old_cols=ego_df.columns
new_cols=[col.upper() for col in old_cols ]

ego_df=(
    ego_df
    .select(
        *[
            F.col(old_cols).alias(new_cols) for ocol, ncol in zip(new_cols,old_cols)
        ]
    )
)
ego_df.display()
```

```python
old_cols=ego_df.columns
new_cols=[col.upper() for col in old_cols]

ego_df=(
    ego_df
    .select(
        *[
            F.col(old_cols).alias(new_cols) for old_cols, new_cols in zip(old_cols,new_cols)
        ]
    )
)
ego_df.display()
```

```python
##ego_df.filter( F.col("PRICE_USD")>50000).select("MODEL","PRICE_USD").display()
##ego_df.select("MODEL").filter(F.col("HORSEPOWER")>400).display()
##ego_df.select("MODEL").where(F.col("AUTOMATIC")==True).display()
##ego_df.filter((F.col("FUEL")=="Petrol")&(F.col("HORSEPOWER")>300)).display()
ego_df=(
    ego_df
     .groupBy("Brand")
    .agg(
        F.count("MODEL").alias("total_cars")
    )
)
ego_df.display()
```

```python
count_car_df=(
    ego_df
    .groupBy("FUEL")
    .agg(
        F.count("MODEL").alias("total_num_cars")
    )
)
count_car_df.display()
```

```python
hema_df=(
    ego_df
    .groupBy("COUNTRY")
    .agg(
        F.count("MODEL").alias("num_cars")
    )
)
hema_df.display()
```

```python
vysh_df=(
    ego_df
    .groupBy("Brand")
    .agg(
        F.avg("PRICE_USD").alias("new_price")
    )
)
vysh_df.display()
```

```python
dhan_df=(
    ego_df
    .groupBy("Brand")
    .agg(
        F.max("HORSEPOWER").alias("maximum_horse")
)
)
dhan_df.display()
```

```python
bhas_df=(
    ego_df
    .groupBy("Brand","Fuel")
    .agg(
        F.count("MODEL").alias("new_cars")
    )
)
bhas_df.display()
```

```python
hge_df=(
    ego_df
    .groupBy("Brand")
    .agg(
        F.count("MODEL").alias("Models")
    )
    .filter(F.col("Models")>3)
)
hge_df.display()
```

```python
klu_df=(
    ego_df
    .groupBy("COUNTRY")
    .agg(
        F.count("MODEL").alias("total_cars")
    )
    .filter(F.col("total_cars")>10)
)
klu_df.display()
```

```python
hytr_df=(
    ego_df
    .groupBy("Brand","Fuel")
    .agg(
        F.count("MODEL").alias("new_cars")
    )
    .filter((F.col("Fuel")=="Petrol")&
    (F.col("new_cars")>2))   
)
hytr_df.display()
```

```python
ert_df=(
    ego_df
    .groupBy("Brand","Fuel")
    .agg(
        F.count("Model").alias("new_cars")
    )
    .filter(F.col("Fuel")=="Petrol")
)
display(ert_df)
```

```python
tyu_df=(
    ego_df
    .groupBy("COUNTRY","MODEL")
    .agg(
        F.avg("Price_USD").alias("new_price")
    )
    .where(F.col("new_price")>50000)
)
display(tyu_df)
```

```python
hiru_df=(
    ego_df
    .groupBy("Brand")
    .agg(
        F.count("automatic").alias("new_autos")
    )
)
hiru_df.display()

    
```

```python
yuti_df=(
    ego_df
    .groupBy("Fuel")
    .agg(
        F.avg("Horsepower").alias("avg_horse")
    )
    .filter(F.col("avg_horse")>300)
)
yuti_df.display()
```
